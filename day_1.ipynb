```python
name = "Bijoy"
fruits = ["aam", "jam", "kathal"]
for f in fruits:
    print(name, "likes", f)
    print(name, "love", f)
```

```python
marks = [70, 85, 90, 84]
total = 0
for m in marks:
    total = total + m
print("Total:", total)
print("Average:", total / len(marks))
```

```python
marks = [70, 85, 90, 45, 95, 30]
for m in marks:
    if m >= 90:
        print(m, "A+")
    if m >= 75:
        print(m, "A")
    elif m >= 50:
        print(m, "Pass")
    else:
        print(m, "Fail")
```
