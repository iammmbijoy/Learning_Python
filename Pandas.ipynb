```python
import pandas as pd

data = {
    "name": ["Bijoy", "Fardin", "Emon"],
    "marks": [70, 85, 90],
    "city": ["Dhaka", "Khulna", "Sylhet"]
}
df = pd.DataFrame(data)
print(df)
print(df["marks"].mean())
print(df[df["marks"] > 75])
print(df["marks"].max())
print(df["marks"].min())
```

```python
df.to_csv("students.csv", index=False)
df2 = pd.read_csv("students.csv")
print(df2)
```
